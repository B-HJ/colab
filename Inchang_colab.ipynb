### 🥨 프레첼 모양 그리기

수학적 매개변수 방정식을 활용하여 프레첼 모양의 곡선을 2차원 그래프로 시각화합니다.

```python
import numpy as np
import matplotlib.pyplot as plt

# 0부터 2*pi까지 정밀한 t 값 생성
t = np.linspace(0, 2 * np.pi, 1000)

# 프레첼 모양을 만드는 매개변수 방정식 정의
x = (2 + np.cos(2 * t)) * np.cos(3 * t)
y = (2 + np.cos(2 * t)) * np.sin(3 * t)

# 그래프 그리기
plt.figure(figsize=(8, 8))
plt.plot(x, y, color='#D2691E', linewidth=5, label='Pretzel Curve') # 프레첼 색상(Chocolate)

# 배경 및 스타일 설정
plt.title('Mathematical Pretzel 🥨', fontsize=16, fontweight='bold', pad=15)
plt.axhline(0, color='gray', linewidth=0.5, linestyle='--')
plt.axvline(0, color='gray', linewidth=0.5, linestyle='--')
plt.grid(True, linestyle=':', alpha=0.6)
plt.axis('equal')
plt.legend(loc='upper right')

# 출력
plt.show()
```

### 🥨 더 진짜 같은 프레첼 그리기 (소금 토핑 추가)

매개변수 곡선의 두께를 키우고, 그 위에 무작위로 하얀 소금 알갱이를 얹어 사실적인 프레첼을 시각화합니다.

```python
import numpy as np
import matplotlib.pyplot as plt

# 프레첼 바디를 위한 매개변수 t
t_val = np.linspace(0, 2 * np.pi, 1000)

# 프레첼 형태를 더 자연스럽게 조정하기 위한 스케일링 함수
x_pretzel = (2.2 + 0.8 * np.cos(2 * t_val)) * np.cos(3 * t_val)
y_pretzel = (1.5 + 0.6 * np.cos(2 * t_val)) * np.sin(3 * t_val) * 0.8

# 그래프 생성
fig, ax = plt.subplots(figsize=(8, 8), facecolor='#FFFDD0') # 따뜻한 크림색 배경

# 1. 프레첼의 어두운 구운 테두리 효과 (가장 아래에 더 두껍게)
ax.plot(x_pretzel, y_pretzel, color='#5C2C16', linewidth=28, solid_capstyle='round')

# 2. 노릇노릇한 황금갈색 메인 빵 부분
ax.plot(x_pretzel, y_pretzel, color='#C16A24', linewidth=22, solid_capstyle='round')

# 3. 빵 윗부분의 밝은 하이라이트 (입체감 부여)
ax.plot(x_pretzel, y_pretzel, color='#E69C5E', linewidth=10, alpha=0.6, solid_capstyle='round')

# 4. 프레첼 위에 뿌려진 하얀 소금 알갱이들 (무작위 분포)
np.random.seed(42)  # 항상 동일하게 예쁜 위치에 뿌려지도록 고정
salt_indices = np.random.choice(len(t_val), size=35, replace=False)
salt_x = x_pretzel[salt_indices] + np.random.normal(0, 0.05, size=35)
salt_y = y_pretzel[salt_indices] + np.random.normal(0, 0.05, size=35)

# 네모난 소금 알갱이 그리기 (흰색 마커, 갈색 테두리)
ax.scatter(salt_x, salt_y, color='white', edgecolor='#8B5A2B', s=90, marker='s', zorder=5, label='Salt Coarse')

# 스타일링
ax.set_title('Delicious Golden Pretzel 🥨', fontsize=18, fontweight='bold', color='#4A2711', pad=20)
ax.set_xlim(-3.5, 3.5)
ax.set_ylim(-2.5, 2.5)
ax.axis('off') # 격자와 축 숨기기

plt.show()
```
